```python
a = np.array([1, 2, 3])
print(a.shape)
print(a.ndim)
print(a.size)
print(a.dtype)
```

```python
b = np.array([[1, 2, 3], [4, 5, 6]])

print(b.shape)
print(b.ndim)
print(b.size)
print(b.dtype)
```

```python
print(a*5)

print(b*5)
```
