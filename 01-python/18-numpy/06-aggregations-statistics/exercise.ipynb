```python
import numpy as np

values = np.array([10, 20, 30, 40, 50])

total = np.sum(values)

print(total)
```

```python
scores = np.array([70, 80, 90, 60, 100])

print("sum:", np.sum(scores))
print("mean:", np.mean(scores))
print("median:", np.median(scores))
print("min:", np.min(scores))
print("max:", np.max(scores))
print("std:", np.std(scores))

X = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("\naxis=0 sum:", np.sum(X, axis=0))
print("axis=1 sum:", np.sum(X, axis=1))
print("axis=0 mean:", np.mean(X, axis=0))
print("axis=1 mean:", np.mean(X, axis=1))

print("\nargmin:", np.argmin(scores))
print("argmax:", np.argmax(scores))
```
