### Mini Project — Temperature Dataset Generator

Create a synthetic dataset containing 30 temperature measurements.

Use NumPy to:

- Generate 30 values between 20°C and 40°C.
- Calculate the average.
- Calculate the minimum.
- Calculate the maximum.
- Convert the values to Fahrenheit.

```python
import numpy as np
```

```python
temperatures = np.linspace(20, 40, 30)
```

```python
print("Temperatures in Celsius:")
print(temperatures)
```

```python
def convert_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32
```

```python
arr = np.arange(0, 100, 10)
print(arr)
```

Q: Why is np.arange() more appropriate for the first problem and np.linspace() more appropriate for the second?

- `np.linspace()` generate random values between defined range.
- `np.arange()` generate numbers in a certain range but here we define step size.

### ⭐ Deep Challenge

Suppose you're simulating the first 24 hours of a day.

```python
hours = np.arange(0, 24, 1)
print("Hours of the day:")
print(hours)
```

```python
temperature = 25 + 5 * np.sin(hours / 24 * 2 * np.pi)
print("Temperatures in Fahrenheit:")
print(convert_to_fahrenheit(temperature))
```

```python
print(np.min(temperature))
```

```python
print(np.max(temperature))
```
