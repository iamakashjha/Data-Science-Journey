## Mini Project
- Student Database

```python
students = {
    101: {
        "name": "Rahul",
        "marks": 88
    },
    102: {
        "name": "Alice",
        "marks": 91
    },
    103: {
        "name": "Bob",
        "marks": 76
    }
}
```

**Tasks:**

- Print all students
- Find the highest marks
- Print the topper
- Add a new student

```python
for student_id, student_info in students.items():
    name = student_info["name"]
    marks = student_info["marks"]
    print(f"Student ID: {student_id}, Name: {name}, Marks: {marks}")
```

```python
# Find the highest marks
highest_marks = max(student_info["marks"] for student_info in students.values())
print(f"The highest marks are: {highest_marks}")
```

```python
# Print the topper
topper = max(students.items(), key=lambda item: item[1]["marks"])
topper_id, topper_info = topper
print(f"The topper is Student ID: {topper_id}, Name: {topper_info['name']}, Marks: {topper_info['marks']}")
```

```python
# Add a new student

new_student_id = 104
new_student_info = {
    "name": "Charlie",
    "marks": 85
}
students[new_student_id] = new_student_info
```

```python
print(students)
```
