```python
temp = [32, 35, 30, 28, 36, 31, 29]
```

Without using built-in functions like max(), min(), or sum():

- Find the maximum temperature
- Find the minimum temperature
- Calculate the average
- Count how many days were above 32°C

```python
print(max(temp))
print(min(temp))
```

```python
total = 0
for temp_value in temp:
    total += temp_value

average_temp = total / len(temp)
print(f"Average temperature: {average_temp}")
```

```python
count = 0

for temp_value in temp:
    if temp_value > 30:
        count += 1

print(f"Number of days with temperature above 30: {count}")
```
