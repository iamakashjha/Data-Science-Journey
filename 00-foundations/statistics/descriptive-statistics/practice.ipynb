# Statistics

Q1: 1. Find mean of:
   10, 20, 30, 40

```python
import statistics
v = [10,20,30,40]

print(statistics.mean(v))
```

Q2: 2. Find median of:
   2, 5, 8, 100

```python
v = [2, 5, 8, 100]

print(statistics.median(v))
```

Q: 3. Find mode of:
   1, 2, 2, 3, 3, 3

```python
v = [1,2,2,3, 3, 3]

print(statistics.mode(v))
```

Q: 4. What is the difference between variance and standard deviation?

- Both tells us the spread or dispersion of the dataset (How far the numbers are from the mean ), the only difference is the unit of measurement **Variance** is the squared difference between the data points from the mean whereas Standard Deviation is the square root difference between the data points from the mean.
