```python
import numpy as np

S = np.array([
    [2, 0],
    [0, 2]
])

v = np.array([
    [2],
    [3]
])

result = S @ v

print(result)
```

### Mini Experiment

```python
import numpy as np

v = np.array([[1], [2]])

S = np.array([
    [3, 0],
    [0, 3]
])

print(S @ v)
```

**Questions:**

### What happened to the vector?
- Vector magnitude has changed.

### Did its direction change?
- No. it's direction is remain unchanged.

### Did its magnitude change?
- Yes, its magnitude has increased.
