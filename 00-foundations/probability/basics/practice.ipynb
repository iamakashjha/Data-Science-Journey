**Difference between theoretical and experimental probability?**

- The theoretical probability is defined or kind of ideal mathmatical logic like if you flip a coin then there are 0.5 chances are if getting either HEAD or TAIL.
- but in experimental probability, we are actually doing the experiment and then calculating the probability. So if you flip a coin 100 times and you get 60 heads and 40 tails then the experimental probability of getting heads is 0.6 and tails is 0.4. So experimental probability is based on actual data and theoretical probability is based on mathematical logic.
