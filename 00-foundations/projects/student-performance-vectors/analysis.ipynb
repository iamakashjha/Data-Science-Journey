### Task 1 — Load the Dataset

```python
import pandas as pd

df = pd.read_csv("dataset.csv")

print(df)
```

### Task 2 — Convert to a Matrix

Extract only the numeric columns.

```python
import numpy as np

X = df[["Math", "Science", "English"]].to_numpy()

print(X)
print(X.shape)
```

**Questions:**

- How many rows?
- How many columns?
- What does each row represent?
- What does each column represent?

```python
print(f'Number of rows in the dataset: {df.shape[0]}')
print(f'Number of columns in the dataset: {df.shape[1]}')
```

### Task 3 — Vector Magnitude

Calculate the magnitude of each student's score vector.

```python
magnitudes = np.linalg.norm(X, axis=1)

print(magnitudes)
```

Interpretation:

- Which student has the largest overall score vector?

### Task 4 — Dot Product

Compare two students.

```python
alice = X[0]
charlie = X[2]

similarity = np.dot(alice, charlie)

print(similarity)
```

Question:

- Does a higher dot product suggest more similar score patterns?

Yes

### Task 5 — Matrix Transformation

Scale every score by 10%.

```python
scaled = X * 1.10

print(scaled)
```

Observe:

- Direction of each score pattern stays the same.
- Magnitude increases.

### Task 6 — Matrix Multiplication

Suppose each subject has a weight:

```python
weights = np.array([
    [0.4],
    [0.35],
    [0.25]
])

final_score = X @ weights

print(final_score)
```

Questions:

- Why does the multiplication work?
- What is the shape of the result?

This is exactly how weighted scoring systems are implemented.

### Task 7 — Eigenvalue Preview

```python
cov_matrix = np.cov(X.T)

eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print(eigenvalues)
print(eigenvectors)
```

Answer:

- How many eigenvalues are returned?
- Why might they be useful?
