## Task 4 — Calculate IQR

Use:
IQR=Q3−Q1

Then determine:  
- Lower bound
- Upper bound

using:  
Q1 − 1.5×IQR  

Q3 + 1.5×IQR  

**Question:**
Is there an outlier?

```python
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)
IQR = Q3 - Q1
print(IQR)
```

```python
print(df[df['Salary'] < (Q1 - 1.5 * IQR)])
```

```python
print(df[df['Salary'] > (Q3 + 1.5 * IQR)])
```

- Yes: there is an outlier in the database which is employee J with of 1000000 salary.
