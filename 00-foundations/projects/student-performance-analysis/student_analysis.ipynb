### Task 1: Load Data

```python
import pandas as pd

df = pd.read_csv("dataset.csv")

print(df)
```

### Task 2: Calculate Statistics

**Find:**
- Mean Marks
- Median Marks
- Maximum Marks
- Minimum Marks

```python
print(df["Marks"].mean())
print(df["Marks"].median())
print(df["Marks"].std())
print(df["Marks"].min())
print(df["Marks"].max())
```

```python
# Plotting the distribution of marks and its mean, median, and standard deviation

plt.hist(df["Marks"], bins=10, alpha=0.5, color="blue", edgecolor="black")
plt.axvline(df["Marks"].mean(), color="red", linestyle="dashed", linewidth=2, label="Mean")
plt.axvline(df["Marks"].median(), color="green", linestyle="dashed", linewidth=2, label="Median")
plt.axvline(df["Marks"].std(), color="orange", linestyle="dashed", linewidth=2, label="Standard Deviation")
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```
