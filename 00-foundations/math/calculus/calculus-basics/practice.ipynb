### 3. Why is optimization important in ML?
- optimization is how a model learns.

##### 1. It Minimizes the Error (Loss Function)
When you initialize an ML model, its internal parameters (weights and biases) are set randomly, meaning its predictions will be wildly inaccurate.

**The Loss Function:** Measures how wrong the model's predictions are compared to the actual data.

**The Optimizer:** Acts as a navigator. Its sole job is to tweak the model's parameters in a calculated way to drive that loss function down as close to zero as possible.

### 5. What is gradient descent trying to do?

**The Goal:** Find the specific weights that make the loss as close to zero as possible.

**The Strategy:** Gradient descent takes iterative steps in the direction of the steepest decrease of the function until it reaches the lowest point (the global minimum or a acceptable local minimum).
